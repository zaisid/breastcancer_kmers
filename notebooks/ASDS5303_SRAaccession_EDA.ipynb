```python
%%bash
# Download the SRA Toolkit release
wget https://ftp-trace.ncbi.nlm.nih.gov/sra/sdk/current/sratoolkit.current-ubuntu64.tar.gz

# Extract the downloaded archive
tar -xzf sratoolkit.current-ubuntu64.tar.gz

# Remove the tar.gz file to save space
rm sratoolkit.current-ubuntu64.tar.gz

# Find the exact directory name (it contains the version number)
SRA_DIR=$(find . -maxdepth 1 -type d -name "sratoolkit*")
echo "SRA Toolkit extracted to: $SRA_DIR"
```

```python
import os

# Find the extracted folder dynamically and add its 'bin' directory to the PATH
sra_dir = [f for f in os.listdir('.') if f.startswith('sratoolkit') and os.path.isdir(f)]
if sra_dir:
    sra_bin_path = os.path.abspath(os.path.join(sra_dir[0], 'bin'))
    os.environ['PATH'] = f"{sra_bin_path}:{os.environ['PATH']}"
    print(f"Added {sra_bin_path} to PATH")
else:
    print("SRA Toolkit directory not found!")
```

If you want to extract/download only a specific range of sequence spots (e.g., reads 1 to 10000) directly to FASTQ format, you can bypass full prefetch and use `fasterq-dump` with the subset flags `-N` (minimum spot) and `-X` (maximum spot):

```python
# Extract only spots 1 to 10,000 from the SRA ID using fastq-dump
!fastq-dump --split-files -N 1 -X 10000 $id
```

```python
from collections import Counter

fastq_path = "/content/SRR41124782_1.fastq"
lengths = []

with open(fastq_path, "r") as f:
    # FASTQ files have 4 lines per record; the sequence is on the 2nd line of each record
    for i, line in enumerate(f):
        if i % 4 == 1:
            lengths.append(len(line.strip()))

# Calculate length distribution
length_counts = Counter(lengths)
print("Read Length Distribution for SRR41124782_1.fastq:")
for length, count in length_counts.items():
    print(f"Length: {length} bp, Count: {count} reads")
```
