```python
%%R
# Create the directory structure expected by SRA_Load_Kmerize.R
dir.create("../breastcancer_kmers/data/", recursive = TRUE, showWarnings = FALSE)

# Copy the files to the expected directory
file.copy("abc_ids.txt", "../breastcancer_kmers/data/abc_ids.txt", overwrite = TRUE)
file.copy("abca_ids.txt", "../breastcancer_kmers/data/abca_ids.txt", overwrite = TRUE)
file.copy("ebc_ids.txt", "../breastcancer_kmers/data/ebc_ids.txt", overwrite = TRUE)
file.copy("bbt_ids.txt", "../breastcancer_kmers/data/bbt_ids.txt", overwrite = TRUE)

# Install Bioconductor manager
if (!requireNamespace("BiocManager", quietly = TRUE)) {
    install.packages("BiocManager")
}

# Install Biostrings
BiocManager::install("Biostrings", update = FALSE, ask = FALSE)

# Load libraries
library(data.table)
library(Biostrings)
library(dplyr)

# Source your functions
source("/content/SRA_Load_Kmerize.R")
```

```python
%%R
for (k in c(2,3,4)) {
    abc_kmers <- kmerize_id_list("abc_ids.txt",
                                 "ABC",k=k,sample_size=2e6,resample=TRUE)

    abca_kmers <- kmerize_id_list("abca_ids.txt",
                                  "ABCA",k=k,sample_size=2e6,resample=TRUE)

    ebc_kmers <- kmerize_id_list("ebc_ids.txt",
                                 "EBC",k=k,sample_size=2e6,resample=TRUE)

    bbt_kmers <- kmerize_id_list("bbt_ids.txt",
                                  "BBT",k=k,sample_size=2e6,resample=TRUE)

    kmer_freqs <- rbind(abc_kmers,abca_kmers,ebc_kmers,bbt_kmers)

    write.csv(kmer_freqs,
              paste0("k", k,
                     "_s2e6_c4_freqs.csv"),
              row.names = TRUE)
}
```
